```python

import json
import pandas as pd

#load firebase json file 

file_path = "wireless-humid-temp-default-rtdb-export 2.json"

with open(file_path, "r", encoding="utf-8") as file:
    firebase_data = json.load(file)

# extracting history data

history_data = firebase_data.get("02_history", {})

if not history_data:
    raise ValueError("No records found in 02_history!")


# 3. CONVERT NESTED JSON INTO ROWS


records = []

for record_id, values in history_data.items():

    records.append({
        "record_id": record_id,
        "humidity": values.get("humidity"),
        "status": values.get("status"),
        "temperature": values.get("temperature"),
        "timestamp": values.get("timestamp")
    })


# 4. CREATE DATAFRAME


df = pd.DataFrame(records)


# 5. DATA TYPE CONVERSION


df["humidity"] = pd.to_numeric(
    df["humidity"], errors="coerce"
)

df["temperature"] = pd.to_numeric(
    df["temperature"], errors="coerce"
)

df["timestamp"] = pd.to_numeric(
    df["timestamp"], errors="coerce"
).astype("Int64")


# 6. SORT BY TIMESTAMP


df = df.sort_values(
    by="timestamp"
).reset_index(drop=True)

# 7. EXPORT CSV AND EXCEL


df.to_csv(
    "DHT22_Dataset.csv",
    index=False
)

#df.to_excel( "DHT22_Dataset.xlsx",index=False)

# 8. DISPLAY DATASET


print("Firebase JSON converted successfully!")
print("Total records:", len(df))

print("\nDataset information:")
df.info()

print("\nFirst 10 records:")
display(df.head(10))

print("\nMissing values:")
print(df.isnull().sum())

print("\nDataset shape:", df.shape)
```
