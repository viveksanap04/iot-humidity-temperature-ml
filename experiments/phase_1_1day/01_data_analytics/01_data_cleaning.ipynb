## 1. Import the libraries

```python
import pandas as pd 
from pathlib import Path
```

```python

# Go up 3 levels from experiments/phase_1_1day/0.../ to the project root
home_ds = Path("../../../datasets/phase_1_1day/processed/")
exact_loc = home_ds / "DHT22_Dataset.csv"
ds = pd.read_csv(exact_loc)
```

## 2.Understand the dataset

```python
# Check column name , data types and basic statistical information

print("\\nData-types")
display(ds.dtypes)
print("-------------------------------------")

display(ds.describe(include="all").T)
print("-------------------------------------")

print("\\nMissing values in each columns:")
display(ds.isnull().sum())
print("-------------------------------------")

print("\\nDuplicate complete rows :",ds.duplicated().sum())
print("Duplicate record IDs:",ds["record_id"].duplicated().sum() if "record_id" in ds else "record_id in column not found")
print("-------------------------------------")

if "status" in ds.columns:
    print("\\nSensor status count:")
    display(ds["status"].value_counts(dropna=False))
```

## 3.Data cleaning

```python
# working on copy so original dataset remains intact
df=ds.copy()

# convert timestamp to Unix milliseconds to Indian standard time 
# utc=True handales original timestamp consistently before converting to IST
df['timestamp']=pd.to_datetime(df['timestamp'],unit='ms',utc=True).dt.tz_convert("Asia/Kolkata")

# Re-sort by timestamp to maintain chronological order regardless of data arrival order
df = df.sort_values('timestamp').reset_index(drop=True)   

# remove duplicates time stamp , keeping the first at time.
df=df.drop_duplicates(subset='timestamp',keep='first') 


```
